# methodology and limitations

## methodology

historical daily adjusted closing prices were collected using the yfinance python package. the study includes eight publicly traded technology companies representing artificial intelligence, semiconductors, software, internet services, and electric vehicles.

the analysis focuses on long-term investment behavior between january 2020 and the present. four primary metrics were calculated:

- average daily return
- volatility
- total return
- maximum drawdown

these measures provide complementary perspectives on performance by examining both growth and investment risk.

## limitations

this project evaluates historical stock performance and does not predict future returns or provide investment advice.

several important limitations should be considered:

- stock prices are influenced by macroeconomic conditions, interest rates, geopolitical events, and company-specific news.
- historical performance does not guarantee future performance.
- the selected companies represent a small sample of the broader technology sector.
- risk is measured primarily through price volatility and maximum drawdown rather than broader financial indicators.

future work could incorporate valuation metrics, earnings growth, financial ratios, or comparisons with market benchmarks such as the s&p 500.
