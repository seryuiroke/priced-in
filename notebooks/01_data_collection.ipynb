## research question

**how do leading ai and technology companies differ in growth, volatility, and long-term investment risk over time?**

this notebook collects and prepares historical market data for eight technology-focused companies.

the project focuses on:

- apple
- microsoft
- nvidia
- alphabet
- meta
- tesla
- amd
- palantir

the analysis will compare market performance using measures of return, volatility, drawdown, and long-term price behavior.

stock performance is analyzed descriptively and is not intended as investment advice or a prediction of future returns.

```python
import yfinance as yf
import pandas as pd
```

```python
tickers = [
    "AAPL",   # Apple
    "MSFT",   # Microsoft
    "NVDA",   # Nvidia
    "GOOGL",  # Alphabet
    "META",   # Meta
    "TSLA",   # Tesla
    "AMD",    # AMD
    "PLTR"    # Palantir
]
```

```python
start_date = "2020-01-01"
end_date = "2025-07-31"
```

```python
stock_data = yf.download(
    tickers,
    start=start_date,
    end=end_date,
    auto_adjust=True,
    group_by="ticker"
)

stock_data.head()
```

```python
stock_data.to_csv("../data/raw/stock_data.csv")

print("stock data saved!")
```
