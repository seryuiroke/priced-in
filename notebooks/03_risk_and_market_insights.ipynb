# risk and market insights

## objective

this notebook interprets the relationship between return and investment risk across eight leading technology companies.

the analysis focuses on three questions:

1. which companies produced the strongest long-term returns?
2. which companies experienced the greatest investment risk?
3. which companies offered the strongest balance between return and risk?

the goal is not to identify the "best" stock or predict future performance. instead, the analysis examines how different technology companies illustrate the economic trade-off between potential reward and uncertainty.

```python
import pandas as pd

summary = pd.read_csv(
    "../data/processed/summary_statistics.csv",
    index_col=0
)

summary
```

```python
best_return = summary["Total Return"].idxmax()

print(f"Highest total return: {best_return}")
print(summary.loc[best_return])
```

```python
highest_risk = summary["Volatility"].idxmax()

print(f"Highest volatility: {highest_risk}")
print(summary.loc[highest_risk])
```

```python
lowest_risk = summary["Volatility"].idxmin()

print(f"Lowest volatility: {lowest_risk}")
print(summary.loc[lowest_risk])
```

## key findings

### highest long-term return

nvidia produced the strongest total return over the study period, substantially outperforming the other companies in the sample.

### highest investment risk

palantir exhibited the greatest volatility and experienced the largest maximum drawdown, indicating greater price uncertainty.

### lowest investment risk

microsoft demonstrated comparatively low volatility while still generating strong long-term returns, illustrating a more balanced risk-return profile.
